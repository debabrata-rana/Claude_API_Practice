### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
claude_model = os.getenv("CLAUDE_MODEL")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=anthropic_api_key)
```

### Calling the Messages API

```python
response = client.messages.create(
    model=claude_model,
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
)
```

### Printing Response

```python
for block in response.content:
    if block.type == "text":
        print(block.text)
```

### Streaming Messages for Better User Experience

```python
with client.messages.stream(
    model=claude_model,
    max_tokens=1000,
    messages=[
        {
            "role": "user",
            "content": "What is quantum computing? Answer in one sentence"
        }
    ]
) as stream:
    for text in stream.text_stream:
        print(text, end="", flush=True)
```

### Passing Assistant Response to the Next Prompt

```python
response = client.messages.create(
    model=claude_model,
    max_tokens=1000,
    system="You are a helpful assistant.",
    messages=[
        {
            "role": "user",
            "content": "How many Claude models are there?"
        },
        {
            "role": "assistant",
            "content": "Claude Haiku, Claude Sonnet, and Claude Opus are the main models, with various versions released over time."
        },
        {
            "role": "user",
            "content": "Can you list the different versions of Claude models?"
        }

    ]
)

for block in response.content:
    if block.type == "text":
        print(block.text)
```
