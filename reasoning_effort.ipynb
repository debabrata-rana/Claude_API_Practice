### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
claude_model = os.getenv("CLAUDE_MODEL")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=anthropic_api_key)
```

### Calling the Messages API with Reasoning Effort

```python
user_prompt = """
You're a senior recruiter in an MNC. You're tasked with hiring a senior mobile architect with GenAI experience.

The technical skills of a candidate is as follows:
- 10+ years of experience in mobile application development in Android
- Proficient in Kotlin, Java, Android SDK, and Jetpack libraries

Suggest technical expertise and skill gaps the candidate has.
"""

print(f"\n\nReasoning effort: {"medium".capitalize()}\n")

response = client.messages.create(
    model=claude_model,
    system="You are a helpful assistant.",
    messages=[
        {"role": "user", "content": user_prompt}
    ],
    output_config={
        "effort": "medium"
    },
    max_tokens=1024
)

for block in response.content:
    if block.type == "text":
        print(block.text)

print("\nUsage stats:", response.usage)
print("\n\n" + "-" * 80 + "\n\n")
```
