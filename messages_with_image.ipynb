### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
claude_model = os.getenv("CLAUDE_MODEL")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=anthropic_api_key)
```

### Calling the Messages API with Image

```python
# NOTE: Currently, the Claude-API-Practice repository is private. So, the image URL is not accessible. Please make the repository public or use a different image URL that is publicly accessible.
image_url = "https://raw.githubusercontent.com/debabrata-rana/Claude-API-Practice/refs/heads/main/Images/nature.jpg"

with client.messages.stream(
    model=claude_model,
    max_tokens=1024,
    system="You are a helpful assistant that answers questions about images.",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "source": {
                        "type": "url",
                        "url": image_url
                    }
                },
                {
                    "type": "text",
                    "text": "Can you find out how many birds are there in the image?"
                }
            ]
        }
    ]
) as stream:
    for text in stream.text_stream:
        print(text, end="", flush=True)
```

```python
import base64
import httpx

# NOTE: Currently, the Claude-API-Practice repository is private. So, the image URL is not accessible. Please make the repository public or use a different image URL that is publicly accessible.
image_url = "https://raw.githubusercontent.com/debabrata-rana/Claude-API-Practice/refs/heads/main/Images/nature.jpg"
image_media_type = "image/jpeg"
image_data = base64.standard_b64encode(
    httpx.get(image_url).content).decode("utf-8")

# passing Base64 encoded image data in the API call
with client.messages.stream(
    model=claude_model,
    max_tokens=1024,
    system="You are a helpful assistant that answers questions about images.",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image",
                    "source": {
                        "type": "base64",
                        "media_type": image_media_type,
                        "data": image_data
                    }
                },
                {
                    "type": "text",
                    "text": "Describe the image in a few sentences."
                }
            ]
        }
    ]
) as stream:
    for text in stream.text_stream:
        print(text, end="", flush=True)
```
