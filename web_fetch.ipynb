### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
claude_model = os.getenv("CLAUDE_MODEL")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key = anthropic_api_key)
```

### Calling the Messages API with Web Fetch Tool

```python
response = client.messages.create(
    model = claude_model,
    max_tokens = 1024,
    messages = [
        {
            "role": "user",
            "content": "What's the weather in Bangalore today as fetched from the URL - https://www.accuweather.com/en/in/bengaluru/204108/weather-forecast/204108?"
        }
    ],
    tools=[{"type": "web_fetch_20250910", "name": "web_fetch", "max_uses": 5}],
)

for block in response.content:
    if block.type == "text":
        print(block.text)
```
