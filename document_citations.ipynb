### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
claude_model = os.getenv("CLAUDE_MODEL")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=anthropic_api_key)
```

### Calling the Messages API with Document Context

```python
with open("./Docs/contextualization.txt", "r") as contextualization_file:
    contextualization_text = contextualization_file.read()

with open("./Docs/embedding.txt", "r") as embedding_file:
    embedding_text = embedding_file.read()

with open("./Docs/generation.txt", "r") as generation_file:
    generation_text = generation_file.read()

with open("./Docs/tokenization.txt", "r") as tokenization_file:
    tokenization_text = tokenization_file.read()

response = client.messages.create(
    model=claude_model,
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "document",
                    "source": {
                        "type": "text",
                        "media_type": "text/plain",
                        "data": contextualization_text
                    },
                    "title": "contextualization.txt",
                    "context": "This document provides an overview of contextualization in AI systems.",
                    "citations": {"enabled": True}
                },
                {
                    "type": "document",
                    "source": {
                        "type": "text",
                        "media_type": "text/plain",
                        "data": embedding_text
                    },
                    "title": "embedding.txt",
                    "context": "This document explains the concept of embeddings in machine learning.",
                    "citations": {"enabled": True}
                },
                {
                    "type": "document",
                    "source": {
                        "type": "text",
                        "media_type": "text/plain",
                        "data": generation_text
                    },
                    "title": "generation.txt",
                    "context": "This document discusses text generation techniques in AI.",
                    "citations": {"enabled": True}
                },
                {
                    "type": "document",
                    "source": {
                        "type": "text",
                        "media_type": "text/plain",
                        "data": tokenization_text
                    },
                    "title": "tokenization.txt",
                    "context": "This document covers tokenization methods used in natural language processing.",
                    "citations": {"enabled": True}
                },
                {
                    "type": "text",
                    "text": "Please summarize the key points from the provided documents and explain how they relate to each other in the context of AI systems."
                }
            ]
        }
    ]
)
```

### Printing Response with Document Citations

```python
print("\n" + "=" * 80)
print("Claude Response")
print("=" * 80)

for block in response.content:

    if block.type != "text":
        continue

    print(block.text)

    if block.citations:
        print("\n📖 Citations:")

        for citation in block.citations:
            print(f"  • Source : {citation.document_title}")
            print(f"    Text   : {citation.cited_text}")

    print()
```
