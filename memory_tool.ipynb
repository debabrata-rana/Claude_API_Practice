### Setting up the Environment

```python
import os
from dotenv import load_dotenv

load_dotenv()

anthropic_api_key = os.getenv("ANTHROPIC_API_KEY")
claude_model = os.getenv("CLAUDE_MODEL")
```

### Creating the Anthropic Client

```python
import anthropic

client = anthropic.Anthropic(api_key=anthropic_api_key)
```

### Configuring the Memory Tool

```python
from anthropic.tools import BetaLocalFilesystemMemoryTool

memory = BetaLocalFilesystemMemoryTool(
    base_path="./memory"
)
```

### Updating Memory

```python
runner = client.beta.messages.tool_runner(
    model=claude_model,
    max_tokens=1024,
    tools=[memory],
    messages=[
        {
            "role": "user",
            "content": """
                        Remember the following information about 'Coroutine Basics'.

                        To create applications that perform multiple tasks at once, a concept known as
                        concurrency, Kotlin uses coroutines. A coroutine is a suspendable computation
                        that lets you write concurrent code in a clear, sequential style. Coroutines
                        can run concurrently with other coroutines and potentially in parallel.

                        On the JVM and in Kotlin/Native, all concurrent code, such as coroutines,
                        runs on threads, managed by the operating system. Coroutines can suspend their
                        execution instead of blocking a thread. This allows one coroutine to suspend
                        while waiting for some data to arrive and another coroutine to run on the same
                        thread, ensuring effective resource utilization.
                        """
        }
    ]
)

runner.until_done()
```

### Accessing Saved Memory

```python
runner = client.beta.messages.tool_runner(
    model=claude_model,
    max_tokens=1024,
    tools=[memory],
    messages=[
        {
            "role": "user",
            "content": "To run program concurrently, what concept does Kotlin use?"
        }
    ]
)

response = runner.until_done()

for block in response.content:
    if block.type == "text":
        print(block.text)
```
